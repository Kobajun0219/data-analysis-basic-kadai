```python
import numpy as np
import matplotlib.pyplot as plt

# 2行3列のサブプロットを作成
fig, axes = plt.subplots(2, 3, figsize=(12, 8), tight_layout=True)

# 1. 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0, marker='o')
axes[0, 0].set_title('Line Plot')

# 2. sin関数
# 0から10までの間で等間隔に100個のデータを作成
x1 = np.linspace(0, 10, 100)
y1 = 2 + 2 * np.sin(2 * x1)
axes[0, 1].plot(x1, y1, linewidth=2.0)
axes[0, 1].set(
    title='Sin Function',
    xlim=(0, 10),
    xticks=np.arange(0, 11, 2),
    ylim=(-1, 5),
    yticks=np.arange(-1, 6)
)

# 3. ヒストグラム
# 平均15, 標準偏差5の正規分布から2000個の乱数を生成
np.random.seed(42)  # 再現性のためのシード固定
x2 = np.random.normal(15, 5, 2000)
axes[0, 2].hist(x2, bins=30, edgecolor='black', alpha=0.7)
axes[0, 2].set_title('Histogram')

# 4. 散布図
# データを生成
np.random.seed(3)
x3 = 4 + np.random.normal(0, 2, 60)
y3 = 4 + np.random.normal(0, 2, len(x3))
axes[1, 0].scatter(x3, y3, alpha=0.8, edgecolors='none')
axes[1, 0].set_title('Scatter Plot')

# 5. 空白の領域（必要に応じて将来用に使用、または非表示にする）
axes[1, 1].set_title('Empty Plot 1')
axes[1, 2].set_title('Empty Plot 2')

# グラフの表示
plt.show()
```
